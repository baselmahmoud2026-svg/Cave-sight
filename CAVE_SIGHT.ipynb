Between attempts, the sites, features, constants and targets changed **after the labelled results had been seen**. This is the "garden of forking paths": blind *ingestion* does not prevent bias if the *method* keeps being adjusted until the expected answer appears.

### What v5.0 does about it
1. **Pre-registration.** The full method (sites, features, targets, constants, hypotheses) is frozen and SHA-256-hashed **before** any data are downloaded (Cell 2).
2. **Two indices, not one.** The *Geology* index (albedo, relief, slope) is reported separately from the *Environment* index (aridity, sunlight, thermal swing, seasonality). The lunar climate proxies describe a *hot desert*, so mixing them into one score rewards deserts regardless of rock type.
3. **No circular validation.** Umm Jirsan is the template the lunar targets were chosen around. It is a sanity check, not evidence, and it is **excluded** from the volcanic-vs-sand test.
4. **Specification curve.** Every analysis choice that appeared in the notebook's history is re-run, and the share of variants supporting each claim is reported (§5).
5. **Site-selection sensitivity.** The four sites tried and later dropped (Lanzarote, Jeju, Newer Volcanics, Amazon) are re-added as a sensitivity set, with QA flags, instead of disappearing.
6. **Bug fixes.** Solar irradiance is **not** divided by 3.6, because the `RE` community already returns kWh/m²/day. Slope uses one definition per run; v4.0 mixed two. The DEM is labelled correctly as Copernicus **GLO-90**, not SRTM.

7. **New in v8.1 — amended pre-registration, declared BEFORE the first live run.** (a) A **third pre-registered target: the lunar south pole (Shackleton rim, the Artemis permanent-base region)**, built from the Diviner polar-temperature literature that the challenge's Resources tab points to. (d) A **multi-criteria suitability matrix** (site × feature × target) and its heat-map, plus a scan-point heat layer on the map, with the pixel-size caveat printed next to it. (e) Three discovery boxes that sat on top of registry sites (El Pinacate, Cima, Medicine Lake) are replaced by provinces with **no registry site inside them** (Zuni-Bandera, Jordan Craters, Payunia) so the discovery scan can actually discover. (f) The `earthaccess` inventory cell (§9b) is on by default — its search needs no login. *Why this is still honest:* the method hash is recomputed and written into every output; the amendment happened before any scan, benchmark or polar data were downloaded, and the §4 hypotheses and the 8 primary sites are unchanged.

## 1 · Pre-registration: freeze the method before touching data

Everything that could be tuned is declared here and hashed. If anything in this cell is edited, the hash changes, and the hash is written into every output file. A reviewer can therefore check that the results were produced by the declared method.

**Provenance of the lunar reference (Mare Tranquillitatis Pit, 8.3355° N, 33.2220° E)**

| Feature | Target | Status | Basis |
|---|---|---|---|
| `Bare_Earth_Albedo` | 0.10 | Literature | Lunar mare basalt normal albedo ≈ 0.07–0.10 |
| `Local_Relief_m` | 45 | Spec choice ⚠ | 45 m is the radar-derived conduit **width** (Carrer et al. 2024). The pit is ~100 m across and ~105 m deep. The 105 m alternative is tested in §5. |
| `Approx_Slope_Deg` | 15 | Design choice | Steep collapse rim |
| `T2M_Range_C` | 25 | Terrestrial proxy | Real lunar swing ~290 K; capped at Earth's upper bound |
| `Precipitation_mm_day` | 0 | Physical | No atmosphere |
| `Solar_kWh_m2_day` | 7.0 | Terrestrial proxy | Earth's upper bound (no lunar atmosphere) |
| `Albedo_Monthly_SD` | 0.005 | Design choice | Aseasonal surface |

The last four targets are what an *ideal hot desert* looks like. That is why they belong in a separate **Environment** index. They answer "how Moon-like are the field-test conditions?", not "is this a lava-tube terrain?".

**Measured thermal facts are carried separately (§6c), not in the z-score vector:** the pit's shaded floor holds a steady ≈17 °C while the surrounding surface swings between about −173 °C and +127 °C (LRO Diviner, Horvath et al. 2022). Earth 2-m *air*-temperature range is not physically comparable to lunar *surface* temperatures, so these numbers anchor the story and the Phase-2 buffering metric instead of being forced into the scaler.

**Third target (v8.1): Lunar South Pole — Shackleton crater rim (≈ 89.7° S, 0° E), the Artemis permanent-base region.**

| Feature | Target | Status | Basis |
|---|---|---|---|

The *cold* itself is **not** in the 7-feature vector (the vector was frozen for the Moon pit and must stay identical across targets for the scores to be comparable). It is carried as a separate, reported-not-scored **cold-context descriptor**: Diviner shows south-polar surface temperatures far below 0 °C, with permanently shadowed floors at ≈ 25–90 K (Paige et al. 2010, *Science* 330). An Earth site passes the descriptor if its POWER annual-mean 2 m air temperature is below 0 °C. LEND hydrogen is Phase 2.

**Benchmark analogs (v8.1, category `Benchmark_Analog`, excluded from H1–H3):** the challenge statement names Atacama, Haughton Crater and Death Valley as the analogs that are used today but "not well characterized". **The 45 m Earth–Moon width match is a literature finding** (Carrer et al. 2024, *Nature Astronomy*: lunar conduit ~45 m wide; Stewart et al. 2024, *PLOS ONE*: Umm Jirsan passages up to ~45 m wide, ~12 m high, 1,481 m mapped). A surface DEM cannot measure either width, so the pipeline does not "validate" this match.

## 2 · Data: NASA POWER climatology and Copernicus GLO-90 elevation

- **NASA POWER** `climatology/point`, `community=RE`, 2001–2020 means. `ALLSKY_SFC_SW_DWN` is already **kWh m⁻² day⁻¹**. The code reads the unit string returned by the API and converts only if it says MJ. (The earlier cell 16 divided by 3.6, and cell 17 then patched this with a `max() < 3` heuristic. Both are removed.)
  - `Bare_Earth_Albedo` = the lowest monthly albedo (snow-free month). `Annual_Albedo` = ANN (used in §5 only). `Albedo_Monthly_SD` = population SD of the 12 months.
- **DEM**: Open-Meteo Elevation = **Copernicus DEM GLO-90 (90 m)**. Two grids are fetched in batched requests:
  - *primary* 3 × 3 nodes at 0.00135° → `Local_Relief_m` (max − min), `Approx_Slope_Deg` = arctan(ΔZ/150 m) (spec metric), `Horn_Slope_Deg` (latitude-corrected).
  - *alt* 5 × 5 nodes at 90 m → `Relief5_m`, `CentralDiff_Slope5_Deg` (the earlier cell-16 method). Used **only** in §5. The v4.0 table mixed the two grids.

```python
# ============================================================
# Cell 4 — HTTP helper (retries, back-off, raw caching)
# ============================================================
SESSION = requests.Session()
SESSION.headers.update({"User-Agent": "CAVE-SIGHT/8.0 (NASA Space Apps 2026 research notebook)"})

def get_json(url, params, cache_name, retries=5, timeout=90):
    path = RAW_DIR / f"{cache_name}.json"
    if USE_CACHE and path.exists():
        return json.loads(path.read_text()), "cache"
    last = None
    for attempt in range(retries):
        try:
            r = SESSION.get(url, params=params, timeout=timeout)
            if r.status_code == 429 or r.status_code >= 500:
                raise requests.HTTPError(f"HTTP {r.status_code}: {r.text[:200]}")
            data = r.json()
            if r.status_code >= 400:
                raise requests.HTTPError(f"HTTP {r.status_code}: {data.get('messages', data)}")
            path.write_text(json.dumps(data, indent=1))
            return data, "live"
        except Exception as e:
            last = e; wait = min(60, 3 * 2 ** attempt)
            print(f"   ↻ {cache_name}: attempt {attempt+1}/{retries} failed ({e}); retry in {wait}s")
            time.sleep(wait)
    raise RuntimeError(f"{cache_name}: failed after {retries} attempts → {last}")
```

```python
# ============================================================
# Cell 5 — NASA POWER features
# ============================================================
POWER_URL = "https://power.larc.nasa.gov/api/temporal/climatology/point"
POWER_PARAMS = METHOD_SPEC["power"]["params"]
MONTHS = ["JAN","FEB","MAR","APR","MAY","JUN","JUL","AUG","SEP","OCT","NOV","DEC"]
MIN_ALBEDO_MONTHS = 6   # v8.1 (was 10): polar-night months are fill values at the polar benchmark sites; all 8 primary sites have 12 months, so their values are unchanged

def _ok(v):
    try: v = float(v)
    except (TypeError, ValueError): return False
    return np.isfinite(v) and v > -900

def fetch_power_features(sid, lat, lon):
    data, src = get_json(POWER_URL, {"parameters": ",".join(POWER_PARAMS), "community": "RE",
                                     "longitude": round(lon, 4), "latitude": round(lat, 4), "format": "JSON"}, f"power_{sid}")
    par = data["properties"]["parameter"]
    units = {k: data.get("parameters", {}).get(k, {}).get("units", "?") for k in POWER_PARAMS}
    alb = [float(par["ALLSKY_SRF_ALB"][m]) for m in MONTHS if _ok(par["ALLSKY_SRF_ALB"].get(m)) and 0 <= float(par["ALLSKY_SRF_ALB"][m]) <= 1]
    ann = {k: float(par[k]["ANN"]) if _ok(par[k].get("ANN")) else np.nan for k in POWER_PARAMS}
    solar, unit = ann["ALLSKY_SFC_SW_DWN"], str(units["ALLSKY_SFC_SW_DWN"])
    if "MJ" in unit.upper(): solar = solar / 3.6
    grid = data.get("geometry", {}).get("coordinates", [None, None])
    f = {"Bare_Earth_Albedo": min(alb) if len(alb) >= MIN_ALBEDO_MONTHS else np.nan,
         "Albedo_Monthly_SD": float(np.std(alb, ddof=0)) if len(alb) >= MIN_ALBEDO_MONTHS else np.nan,
         "Albedo_N_Months": len(alb),
         "Annual_Albedo": ann["ALLSKY_SRF_ALB"], "T2M_Mean_C": ann["T2M"], "T2M_Range_C": ann["T2M_RANGE"],
         "Precipitation_mm_day": ann["PRECTOTCORR"], "Solar_kWh_m2_day": solar,
         "Solar_Unit_Reported": unit, "POWER_Grid_Lon": grid[0], "POWER_Grid_Lat": grid[1], "POWER_Source": src}
    core = ["Bare_Earth_Albedo", "Albedo_Monthly_SD", "T2M_Range_C", "Precipitation_mm_day", "Solar_kWh_m2_day"]
    f["POWER_Status"] = "Success" if all(np.isfinite(f[k]) for k in core) else "Incomplete"
    return f

rows = []
for _, s in BLIND_REGISTRY.iterrows():
    try: f = fetch_power_features(s.Site_ID, s.Lat, s.Lon)
    except Exception as e: f = {"POWER_Status": f"Failed: {e}"}; print(f"✗ {s.Site_ID}: {e}")
    f["Site_ID"] = s.Site_ID; rows.append(f)
    if f.get("POWER_Status") == "Success":
        print(f"✓ {s.Site_ID} [{f['POWER_Source']}] bare_alb={f['Bare_Earth_Albedo']:.3f} solar={f['Solar_kWh_m2_day']:.2f} ({f['Solar_Unit_Reported']})")
    if f.get("POWER_Source") != "cache": time.sleep(0.4)
POWER_DF = pd.DataFrame(rows).set_index("Site_ID")
```

```python
# ============================================================
# Cell 6 — DEM grids (batched) + morphometry
# ============================================================
DEM_URL, DEM_SOURCE = "https://api.open-meteo.com/v1/elevation", "Copernicus DEM GLO-90 (90 m) via Open-Meteo"
M_PER_DEG = 111_320.0

def grid_nodes(lat, lon, n, step_lat_deg, step_lon_deg):
    k = (n - 1) // 2   # rows north → south, cols west → east
    return [(round(lat + r * step_lat_deg, 6), round(lon + c * step_lon_deg, 6)) for r in range(k, -k - 1, -1) for c in range(-k, k + 1)]

def site_grids(lat, lon):
    p = DEM["primary"]; a = DEM["alt"]
    g3 = grid_nodes(lat, lon, p["n"], p["step_deg"], p["step_deg"])
    s5 = a["step_m"] / M_PER_DEG
    g5 = grid_nodes(lat, lon, a["n"], s5, s5 / math.cos(math.radians(lat)))
    return {"g3": g3, "g5": g5}

def fetch_all_grids(registry):
    plan = [(s.Site_ID, key, pt) for _, s in registry.iterrows() for key, pts in site_grids(s.Lat, s.Lon).items() for pt in pts]
    elev = []
    for i in range(0, len(plan), 100):          # Open-Meteo: max 100 points per request
        chunk = plan[i:i + 100]
        data, _ = get_json(DEM_URL, {"latitude": ",".join(str(p[2][0]) for p in chunk),
                                     "longitude": ",".join(str(p[2][1]) for p in chunk)}, f"dem_batch_{i // 100:02d}")
        vals = data.get("elevation", [])
        if len(vals) != len(chunk): raise ValueError(f"DEM returned {len(vals)} values for {len(chunk)} points")
        elev += vals
    out = {}
    for (sid, key, _), z in zip(plan, elev):
        out.setdefault(sid, {}).setdefault(key, []).append(np.nan if z is None else float(z))
    return {sid: {"g3": np.array(d["g3"]).reshape(3, 3), "g5": np.array(d["g5"]).reshape(5, 5)} for sid, d in out.items()}

def morph3(Z, lat):
    dy = DEM["primary"]["step_deg"] * M_PER_DEG; dx = dy * math.cos(math.radians(lat))
    relief = float(np.nanmax(Z) - np.nanmin(Z))
    a, b, c = Z[0]; d, e, f = Z[1]; g, h, i = Z[2]
    dzdx = ((c + 2*f + i) - (a + 2*d + g)) / (8 * dx); dzdy = ((g + 2*h + i) - (a + 2*b + c)) / (8 * dy)
    return {"Center_Elevation_m": float(e), "Local_Relief_m": relief,
            "Approx_Slope_Deg": math.degrees(math.atan(relief / 150.0)),
            "Horn_Slope_Deg": math.degrees(math.atan(math.hypot(dzdx, dzdy)))}

def morph5(Z):
    s = DEM["alt"]["step_m"]
    dzdx = (Z[2, 3] - Z[2, 1]) / (2 * s); dzdy = (Z[1, 2] - Z[3, 2]) / (2 * s)
    return {"Relief5_m": float(np.nanmax(Z) - np.nanmin(Z)), "CentralDiff_Slope5_Deg": math.degrees(math.atan(math.hypot(dzdx, dzdy)))}

try:
    GRIDS = fetch_all_grids(BLIND_REGISTRY)
except Exception as e:
    print(f"✗ DEM fetch failed: {e}"); GRIDS = {}

rows = []
for _, s in BLIND_REGISTRY.iterrows():
    g = GRIDS.get(s.Site_ID)
    if g is not None and np.isfinite(g["g3"]).all() and np.isfinite(g["g5"]).all():
        r = {**morph3(g["g3"], s.Lat), **morph5(g["g5"]), "DEM_Status": "Success"}
    else:
        r = {"DEM_Status": "Failed"}
    r.update({"Site_ID": s.Site_ID, "DEM_Source": DEM_SOURCE}); rows.append(r)
DEM_DF = pd.DataFrame(rows).set_index("Site_ID")
DEM_DF.round(2)
```

```python
# ============================================================
# Cell 7 — Data-quality gate (no imputation)
# ============================================================
SCORED_FEATURES = sorted({f for fs in PILLARS.values() for f in fs})
EXTRA = ["Annual_Albedo", "Horn_Slope_Deg", "Relief5_m", "CentralDiff_Slope5_Deg", "T2M_Mean_C", "Center_Elevation_m"]
FT = BLIND_REGISTRY.set_index("Site_ID").join(POWER_DF, how="left").join(DEM_DF, how="left")
for c in SCORED_FEATURES + EXTRA:
    if c not in FT: FT[c] = np.nan
SANITY = {"Bare_Earth_Albedo": (0.02, 0.9), "Albedo_Monthly_SD": (0, 0.5), "T2M_Range_C": (0, 40), "Precipitation_mm_day": (0, 30),
          "Solar_kWh_m2_day": (1.5, 9.5), "Local_Relief_m": (0, 2000), "Approx_Slope_Deg": (0, 90)}
QA = pd.DataFrame({"Set": ["primary" if s in PRIMARY_SITE_IDS else "benchmark" if s in BENCHMARK_SITE_IDS else "sensitivity" for s in FT.index],
                   "POWER": FT["POWER_Status"], "DEM": FT["DEM_Status"],
                   "Missing": FT[SCORED_FEATURES + EXTRA].isna().sum(axis=1),
                   "Albedo_Months": FT["Albedo_N_Months"] if "Albedo_N_Months" in FT else np.nan,
                   "Out_of_range": [",".join(k for k, (lo, hi) in SANITY.items() if np.isfinite(r[k]) and not lo <= r[k] <= hi) or "-"
                                    for _, r in FT.iterrows()]}, index=FT.index)
prim = FT.loc[PRIMARY_SITE_IDS, SCORED_FEATURES]
MISSING_PCT = 100 * prim.isna().sum().sum() / prim.size
COMPLETE = QA.query("Missing == 0").index.tolist()
PRIMARY = [s for s in PRIMARY_SITE_IDS if s in COMPLETE]
ALL12   = [s for s in PRIMARY_SITE_IDS + SENSITIVITY_SITE_IDS if s in COMPLETE]
BENCH   = [s for s in BENCHMARK_SITE_IDS if s in COMPLETE]        # v8.1 benchmark analogs with complete data
print(f"Primary set missing measurements: {MISSING_PCT:.1f}% | complete primary sites: {len(PRIMARY)}/8 | complete overall: {len(ALL12)}/12 | benchmark complete: {len(BENCH)}/4")
if len(BENCH) < 4: print("   ⚠ a benchmark site is incomplete (polar DEM/POWER gap?) — it is reported as missing, never imputed")
print(QA.to_string())
```

## 3 · Scoring engine (white-box, label-blind)

z-scores (fitted on the analysed sites + Moon) → per-pillar RMS standardized distance *d* → S = 100·exp(−k·d) → indices:
- **Composite** = geometric mean of Regolith, Topography and Environment (the v4.0 design, kept as the primary model).
- **Geology index** = geometric mean of Regolith and Topography. This is the rock and terrain signal only.
- **Environment index** = Environment pillar. This is how Moon-like the operating climate is. It is descriptive and is not evidence of lava tubes.

```python
# ============================================================
# Cell 8 — Scoring engine
# ============================================================
def score_sites(X, lunar_ref=LUNAR_REF, pillars=PILLARS, decay=DECAY, scaler="std"):
    assert not LABEL_COLUMNS & set(X.columns), "Scoring must be label-blind"
    cols = list(dict.fromkeys(f for fs in pillars.values() for f in fs))
    M = pd.concat([X[cols], pd.DataFrame([{c: lunar_ref[c] for c in cols}], index=["__MOON__"])])
    S = (StandardScaler() if scaler == "std" else MinMaxScaler()).fit_transform(M.values)
    Z = pd.DataFrame(S, index=M.index, columns=cols)
    dz = Z.drop(index="__MOON__") - Z.loc["__MOON__"]
    out = pd.DataFrame(index=X.index)
    for p, fs in pillars.items():
        out[f"{p}_Score_%"] = 100 * np.exp(-decay[p] * np.sqrt((dz[fs] ** 2).mean(axis=1)))
    gm = lambda ps: np.exp(np.log(out[[f"{p}_Score_%" for p in ps if p in pillars]]).mean(axis=1))
    out["Composite_Score_%"] = gm(list(pillars))
    out["Geology_Index_%"] = gm(GEOLOGY_PILLARS)
    out["Environment_Index_%"] = gm(ENVIRONMENT_PILLARS) if "Environment" in pillars else np.nan
    return out, dz

X_PRIMARY = FT.loc[PRIMARY, SCORED_FEATURES].astype(float)
SCORES, DZ = score_sites(X_PRIMARY)

def score_against_reference(X_new, target_ref, ref_X=None, pillars=PILLARS, decay=DECAY):
    """v8.1 frozen reference scaling. The scaler is fitted ONCE on the primary sites + the target row and then
    applied to X_new, so scan points and benchmark sites land on the same z-scale as the primary ranking.
    For X_new == primary sites this reproduces score_sites() exactly (asserted below)."""
    assert not LABEL_COLUMNS & set(X_new.columns), "Scoring must be label-blind"
    ref_X = X_PRIMARY if ref_X is None else ref_X
    cols = list(dict.fromkeys(f for fs in pillars.values() for f in fs))
    t = pd.DataFrame([{c: target_ref[c] for c in cols}], index=["__TARGET__"])
    sc = StandardScaler().fit(pd.concat([ref_X[cols], t]).values)
    Zt = pd.Series(sc.transform(t.values)[0], index=cols)
    Zn = pd.DataFrame(sc.transform(X_new[cols].values), index=X_new.index, columns=cols)
    dz = Zn - Zt
    out = pd.DataFrame(index=X_new.index)
    for p, fs in pillars.items():
        out[f"{p}_Score_%"] = 100 * np.exp(-decay[p] * np.sqrt((dz[fs] ** 2).mean(axis=1)))
    gm = lambda ps: np.exp(np.log(out[[f"{p}_Score_%" for p in ps if p in pillars]]).mean(axis=1))
    out["Composite_Score_%"] = gm(list(pillars)); out["Geology_Index_%"] = gm(GEOLOGY_PILLARS)
    out["Environment_Index_%"] = gm(ENVIRONMENT_PILLARS) if "Environment" in pillars else np.nan
    return out, dz

_chk, _ = score_against_reference(X_PRIMARY, LUNAR_REF)
assert np.allclose(_chk["Composite_Score_%"], SCORES["Composite_Score_%"]), "reference scaling must reproduce the primary scores"
print("✓ frozen reference scaling reproduces the primary Moon scores exactly")
SCORES.round(2)
```

## 4 · Lock ranking → unseal labels → test the pre-declared hypotheses

The test compares the **5 candidates with the 2 sand controls**. The anchor is excluded because it is the template. With n = 5 vs 2, the smallest possible exact one-sided p is 1/21 ≈ 0.048. *p* < 0.05 is therefore reached only when **every** candidate beats both controls. Read it as a perfect-separation test with very low power.

```python
# ============================================================
# Cell 9 — Ranking lock + hypothesis tests
# ============================================================
RANKED = SCORES.sort_values("Composite_Score_%", ascending=False).copy()
RANKED.insert(0, "Rank", range(1, len(RANKED) + 1))
RANK_LOCK = ";".join(f"{s}:{v:.6f}" for s, v in RANKED["Composite_Score_%"].items())
RANK_HASH = hashlib.sha256((SPEC_HASH + RANK_LOCK).encode()).hexdigest()
print(f"🔒 Ranking locked before unsealing labels · sha256 = {RANK_HASH[:24]}…")

LABELS = _SEALED.set_index("Site_ID")           # ← first use of the labels
AUDIT = RANKED.join(LABELS).join(BLIND_REGISTRY.set_index("Site_ID")[["Site_Name", "Lat", "Lon"]])

def separation_test(scores, cats):
    cand = scores[cats == "Volcanic_Candidate"]; ctrl = scores[cats == "Negative_Control"]
    if len(cand) == 0 or len(ctrl) == 0: return {"pass": False, "p": np.nan, "perfect": False, "detail": "insufficient groups"}
    u, p = mannwhitneyu(cand, ctrl, alternative="greater", method="exact")
    perfect = bool(cand.min() > ctrl.max())
    beaten = int((cand < ctrl.max()).sum())
    return {"pass": bool(p < ALPHA and perfect), "p": float(p), "U": float(u), "perfect": perfect,
            "detail": f"U={u:.0f}, exact one-sided p={p:.4f} (min possible {1/math.comb(len(cand)+len(ctrl), len(ctrl)):.4f}); "
                      f"{beaten}/{len(cand)} candidates score below the best control"}

cats = AUDIT["PostHoc_Category"]
anchor_rank = int(AUDIT.loc["SITE_01", "Rank"]) if "SITE_01" in AUDIT.index else None
CHECKS = {
    "H0_sanity":  {"pass": anchor_rank == 1, "detail": f"Umm Jirsan composite rank = {anchor_rank} (sanity only)"},
    "H1_primary": separation_test(AUDIT["Composite_Score_%"], cats),
    "H2":         separation_test(AUDIT["Geology_Index_%"], cats),
    "H3":         separation_test(AUDIT["Regolith_Score_%"], cats),
    "QA_zero_missing": {"pass": MISSING_PCT == 0, "detail": f"missing = {MISSING_PCT:.1f}%"},
}
print("\nPRE-DECLARED HYPOTHESES")
for k, v in CHECKS.items():
    print(f"  {'✅ PASS' if v['pass'] else '❌ FAIL'}  {k:16s} {v['detail']}")

AUDIT[["Rank", "Site_Name", "PostHoc_Category", "Composite_Score_%", "Geology_Index_%", "Environment_Index_%",
       "Regolith_Score_%", "Topography_Score_%"]].round(2)
```

## 6 · Measurement uncertainty and site-selection sensitivity
- **Monte Carlo** (primary model, primary sites). Assumed 1σ: albedo ±0.01, albedo SD ±0.003, T-range ±0.5 °C, precipitation ×LogN(0, 0.10), solar ±3 %, each DEM node ±2 m (relief and slopes recomputed). The full pipeline re-runs on every draw. *These σ are stated assumptions; edit `MC_SIGMA` to change them.*
- **Site selection**: the primary model re-scored on all 12 sites, and on the 10 that remain after removing the island sites, which carry an ocean-pixel risk.

```python
# ============================================================
# Cell 12 — Monte Carlo + site-selection sensitivity
# ============================================================
rng = np.random.default_rng(RANDOM_SEED)
MC_SIGMA = {"Bare_Earth_Albedo": 0.01, "Albedo_Monthly_SD": 0.003, "T2M_Range_C": 0.5, "precip_logsd": 0.10, "solar_frac": 0.03, "dem_node_m": 2.0}
LAT = BLIND_REGISTRY.set_index("Site_ID")["Lat"]

def perturb(X):
    Xp = X.copy(); n = len(Xp)
    Xp["Bare_Earth_Albedo"] = np.clip(Xp["Bare_Earth_Albedo"] + rng.normal(0, MC_SIGMA["Bare_Earth_Albedo"], n), 0.01, None)
    Xp["Albedo_Monthly_SD"] = np.clip(Xp["Albedo_Monthly_SD"] + rng.normal(0, MC_SIGMA["Albedo_Monthly_SD"], n), 0, None)
    Xp["T2M_Range_C"] += rng.normal(0, MC_SIGMA["T2M_Range_C"], n)
    Xp["Precipitation_mm_day"] *= rng.lognormal(0, MC_SIGMA["precip_logsd"], n)
    Xp["Solar_kWh_m2_day"] *= 1 + rng.normal(0, MC_SIGMA["solar_frac"], n)
    for sid in Xp.index:
        m = morph3(GRIDS[sid]["g3"] + rng.normal(0, MC_SIGMA["dem_node_m"], (3, 3)), LAT[sid])
        Xp.at[sid, "Local_Relief_m"], Xp.at[sid, "Approx_Slope_Deg"] = m["Local_Relief_m"], m["Approx_Slope_Deg"]
    return Xp

mc_rank, mc_score, mc_sep_comp, mc_sep_geo = [], [], [], []
cat_p = LABELS.loc[PRIMARY, "PostHoc_Category"]
for _ in range(N_MONTE_CARLO):
    s, _ = score_sites(perturb(X_PRIMARY))
    mc_score.append(s["Composite_Score_%"]); mc_rank.append(s["Composite_Score_%"].rank(ascending=False, method="min"))
    for col, store in [("Composite_Score_%", mc_sep_comp), ("Geology_Index_%", mc_sep_geo)]:
        v = s[col]; store.append(v[cat_p == "Volcanic_Candidate"].min() > v[cat_p == "Negative_Control"].max())
MC_RANKS, MC_SCORES = pd.DataFrame(mc_rank), pd.DataFrame(mc_score)
MC_SUMMARY = pd.DataFrame({"P_Rank1_%": (MC_RANKS == 1).mean() * 100, "Median_Rank": MC_RANKS.median(),
    "Rank_90CI": [f"{int(MC_RANKS[c].quantile(.05))}–{int(MC_RANKS[c].quantile(.95))}" for c in MC_RANKS],
    "Score_5pct": MC_SCORES.quantile(.05), "Score_95pct": MC_SCORES.quantile(.95)}).loc[RANKED.index]
P_SEP = {"composite": 100 * np.mean(mc_sep_comp), "geology": 100 * np.mean(mc_sep_geo)}
print(f"Monte Carlo ({N_MONTE_CARLO}): P(perfect separation) composite = {P_SEP['composite']:.1f}% | geology = {P_SEP['geology']:.1f}%")

SITE_SENS = []
for name, ids in [("primary_8", PRIMARY), ("all_12", ALL12), ("all_12_minus_islands", [s for s in ALL12 if s not in ("SITE_09", "SITE_10")])]:
    s, _ = score_sites(FT.loc[ids, SCORED_FEATURES].astype(float))
    for col in ["Composite_Score_%", "Geology_Index_%"]:
        t = separation_test(s[col], LABELS.loc[ids, "PostHoc_Category"])
        SITE_SENS.append({"site_set": name, "n": len(ids), "index": col, "anchor_rank": int(s[col].rank(ascending=False)["SITE_01"]),
                          "top_site": s[col].idxmax(), "mwu_p": round(t["p"], 4), "detail": t["detail"]})
SITE_SENS = pd.DataFrame(SITE_SENS)
SITE_SENS
```

## 6c · Measured lunar thermal context → the Phase-2 "thermal buffering ratio"

What makes a cave a shelter is measurable from orbit on all three worlds: its entrance temperature swings far less than the surrounding surface (warm at night, cool at midday — the standard thermal cave-detection signature; Wynne et al.). On the Moon, Diviner shows the Mare Tranquillitatis Pit's shaded floor steady at ≈17 °C while the surface swings ≈300 K.

**Definition (pre-declared for Phase 2):** `B = ΔT_diurnal(entrance) / ΔT_diurnal(surrounding surface)`, from midday + predawn thermal imagery (ECOSTRESS / Landsat TIR / airborne on Earth; Diviner on the Moon; THEMIS on Mars). B ≪ 1 ⇒ sheltering void. This is the metric that will let CaveSight test *"tube vs no tube"* against basalt-without-caves controls (e.g. the USGS Pisgah airborne thermal dataset), instead of *"basalt vs sand"*.

```python
# ============================================================
# Cell 13c — Lunar thermal context constants (published) + buffering ratio
# ============================================================
THERMAL_CONTEXT = {
    "moon_surface_day_C": 127.0, "moon_surface_night_C": -173.0,          # LRO Diviner (Horvath et al. 2022)
    "moon_pit_shadow_steady_C": 17.0,                                      # MTP shaded floor, steady through the lunar day
    "moon_pit_shadow_fluctuation_C": 10.0,                                 # "fluctuates only slightly" — conservative upper bound
    "source": "Horvath, Hayne & Paige 2022, Geophys. Res. Lett., doi:10.1029/2022GL099710",
    "buffering_ratio_definition": "B = diurnal_swing(entrance) / diurnal_swing(surface); B<<1 => sheltering void",
}
moon_surface_swing = THERMAL_CONTEXT["moon_surface_day_C"] - THERMAL_CONTEXT["moon_surface_night_C"]
B_moon_pit = THERMAL_CONTEXT["moon_pit_shadow_fluctuation_C"] / moon_surface_swing
THERMAL_CONTEXT["moon_surface_diurnal_swing_K"] = moon_surface_swing
THERMAL_CONTEXT["moon_pit_buffering_ratio_upper_bound"] = round(B_moon_pit, 4)
print(f"Lunar surface diurnal swing ≈ {moon_surface_swing:.0f} K | MTP shaded floor ≈ steady 17 °C → buffering ratio B ≤ {B_moon_pit:.3f}")
print("Phase-2 task: compute the same B for Umm Jirsan and for basalt-without-caves controls from midday+predawn thermal imagery.")
```

Live WMTS tiles from **NASA Moon Trek and Mars Trek** are fetched for the website's dual-planet view (best-effort: a failure is recorded, never fatal).

```python
# ============================================================
# Cell 13e — Mars target scoring + NASA Trek WMTS tile fetch
# ============================================================
MARS_SCORES, _mars_dz = score_against_reference(X_PRIMARY, MARS_REF)   # identical to score_sites(..., MARS_REF) for the primary set
DUAL = pd.DataFrame({
    "Moon_Composite_%": SCORES["Composite_Score_%"], "Mars_Composite_%": MARS_SCORES["Composite_Score_%"],
    "Moon_Geology_%": SCORES["Geology_Index_%"],     "Mars_Geology_%": MARS_SCORES["Geology_Index_%"],
}).join(LABELS["PostHoc_Category"]).join(BLIND_REGISTRY.set_index("Site_ID")["Site_Name"])
DUAL["Better_Fit"] = np.where(DUAL["Mars_Composite_%"] > DUAL["Moon_Composite_%"], "Mars", "Moon")
DUAL = DUAL.sort_values("Mars_Composite_%", ascending=False)
print("Dual-planet ranking (same engine, two pre-registered targets):")
display(DUAL.round(1))

MARS_REGISTRY_NOTE = ("Arsia skylights ('Seven Sisters': Dena, Chloe, Wendy, Annie, Abby, Nikki, Jeanne) — "
                      "exact coordinates and confidence ratings are in the USGS MGC3 catalog; "
                      "this notebook pre-registers the family-level target and leaves per-skylight "
                      "morphometry to Phase 2 with MGC3 + CTX DTMs.")

# --- best-effort NASA Trek tiles (zoom 0 global tile per body) ---
TREK_TILES = {
    "moon_lola_shade": "https://trek.nasa.gov/tiles/Moon/EQ/LRO_LOLA_ClrShade_Global_128ppd_v04/1.0.0/default/default028mm/0/0/0.jpg",
    "moon_wac_mosaic": "https://trek.nasa.gov/tiles/Moon/EQ/LRO_WAC_Mosaic_Global_303ppd_v02/1.0.0/default/default028mm/0/0/0.jpg",
    "mars_mola_shade": "https://trek.nasa.gov/tiles/Mars/EQ/Mars_MGS_MOLA_ClrShade_merge_global_463m/1.0.0/default/default028mm/0/0/0.jpg",
    "mars_viking":     "https://trek.nasa.gov/tiles/Mars/EQ/Mars_Viking_MDIM21_ClrMosaic_global_232m/1.0.0/default/default028mm/0/0/0.jpg",
}
TREK_STATUS = {}
for name, url in TREK_TILES.items():
    try:
        r = SESSION.get(url, timeout=30)
        ok = getattr(r, "status_code", 0) == 200 and getattr(r, "content", b"")[:3] in (b"\xff\xd8\xff", b"\x89PN")
        if ok: (RAW_DIR / f"trek_{name}.jpg").write_bytes(r.content)
        TREK_STATUS[name] = "fetched" if ok else f"unexpected response (HTTP {getattr(r, 'status_code', '?')})"
    except Exception as e:
        TREK_STATUS[name] = f"unavailable here ({type(e).__name__})"
print("NASA Trek tile fetch:", TREK_STATUS)
```

## 6e2 · Three targets, one engine: Moon pit · Mars skylights · Lunar south pole — plus the benchmark analogs (v8.1)

Every site now has **three scores on one z-scale** (frozen reference scaling). The Moon pit and Mars skylight columns answer the cave-shelter question; the **south-pole column answers the permanent-base question** the challenge title asks. The four **benchmark analogs** (Atacama, Haughton Crater, McMurdo Dry Valleys, Death Valley) are the places mission teams use today; they are scored here so a judge can see where the classics sit next to the lava-tube candidates — and they are **excluded from H1–H3**, because one site per environment is not a control group.

The **cold-context descriptor** is printed beside the polar score and is *not* inside it: a site can match the polar target's proxies (low Sun, no rain, small swing, bright regolith) and still be warm. Reporting the two separately is more honest than hiding a temperature term inside a composite.

```python
# ============================================================
# Cell 13e2 — Three-target comparison table + benchmark analogs (v8.1)
# ============================================================
POLAR_SCORES, _polar_dz = score_against_reference(X_PRIMARY, POLAR_REF)
X_BENCH = FT.loc[BENCH, SCORED_FEATURES].astype(float) if len(BENCH) else X_PRIMARY.iloc[0:0]
BENCH_SC = {t: score_against_reference(X_BENCH, ref)[0] if len(BENCH) else None for t, (ref, _) in TARGETS.items()}
PRIM_SC  = {"Moon": SCORES, "Mars": MARS_SCORES, "Polar": POLAR_SCORES}

def _triple_rows(ids, sc):
    df = pd.DataFrame(index=ids)
    for t in TARGETS:
        df[f"{t}_Composite_%"] = sc[t]["Composite_Score_%"]; df[f"{t}_Geology_%"] = sc[t]["Geology_Index_%"]
        df[f"{t}_Environment_%"] = sc[t]["Environment_Index_%"]
    return df

TRIPLE = pd.concat([_triple_rows(PRIMARY, PRIM_SC)] + ([_triple_rows(BENCH, BENCH_SC)] if len(BENCH) else []))
TRIPLE["Best_Target"] = TRIPLE[[f"{t}_Composite_%" for t in TARGETS]].idxmax(axis=1).str.replace("_Composite_%", "")
TRIPLE["T2M_Mean_C"] = FT.loc[TRIPLE.index, "T2M_Mean_C"]
TRIPLE["Cold_Context_Pass"] = TRIPLE["T2M_Mean_C"] < 0            # reported, not scored (see §1)
TRIPLE = TRIPLE.join(LABELS["PostHoc_Category"]).join(BLIND_REGISTRY.set_index("Site_ID")["Site_Name"])
TRIPLE = TRIPLE.sort_values("Moon_Composite_%", ascending=False)

print("THREE-TARGET COMPARISON (one z-scale; benchmark analogs scored but never tested)")
display(TRIPLE[["Site_Name", "PostHoc_Category", "Moon_Composite_%", "Mars_Composite_%", "Polar_Composite_%",
                "Best_Target", "T2M_Mean_C", "Cold_Context_Pass"]].round(1))
if len(BENCH):
    pb = TRIPLE.loc[BENCH].sort_values("Polar_Composite_%", ascending=False)
    print(f"\nPolar-target leader among benchmarks: {pb.iloc[0]['Site_Name']} ({pb.iloc[0]['Polar_Composite_%']:.1f} %)"
          f" · cold-context pass: {', '.join(pb.index[pb.Cold_Context_Pass].map(lambda s: FT.at[s, 'Site_Name'])) or 'none'}")
    lead_lava = TRIPLE.loc[PRIMARY].sort_values("Polar_Composite_%", ascending=False).iloc[0]
    print(f"Best lava-tube site on the polar target: {lead_lava['Site_Name']} ({lead_lava['Polar_Composite_%']:.1f} %) — "
          "if a benchmark beats it, say so: the lava-tube sites are cave analogs, not polar-base analogs.")
print(f"\nRule: {COLD_CONTEXT_RULE}")
```

## 6f · Discovery mode — identifying NEW terrestrial analogs

This is the half of the challenge the fixed 8-site registry cannot answer: *identify new terrestrial analogs*. The protocol was frozen in §1 **before** any scan data was fetched: a deterministic 2 × 3 grid inside each of 7 volcanic provinces and 2 sand scan-controls (54 points), screened **unlabelled** by the same engine, then flagged by the pre-registered novelty rule (≥ 25 km from every catalogued site AND Geology index ≥ the median of the primary volcanic candidates). Scan points are *candidates for field verification*, not confirmed caves — a 55–110 km climate pixel and a 90 m DEM cannot see a tube (§6c's buffering ratio is the Phase-2 follow-up for exactly these points).

```python
# ============================================================
# Cell 13f — Pre-registered grid scan + novelty flagging
# ============================================================
def haversine_km(lat1, lon1, lat2, lon2):
    rl1, rl2 = math.radians(lat1), math.radians(lat2)
    a = math.sin(math.radians(lat2 - lat1) / 2) ** 2 + math.cos(rl1) * math.cos(rl2) * math.sin(math.radians(lon2 - lon1) / 2) ** 2
    return 2 * 6371.0 * math.asin(math.sqrt(a))

def build_scan_points():
    pts = []
    for reg, box in SCAN_REGIONS.items():
        las = np.linspace(box["lat"][0], box["lat"][1], 2 + 2)[1:-1]      # 2 interior rows
        los = np.linspace(box["lon"][0], box["lon"][1], 3 + 2)[1:-1]      # 3 interior cols
        for i, (la, lo) in enumerate([(a, b) for a in las for b in los][:N_SCAN_PER_REGION]):
            pts.append({"Scan_ID": f"SCAN_{reg}_{i+1:02d}", "Region": reg, "Region_Kind": box["kind"],
                        "Lat": round(float(la), 4), "Lon": round(float(lo), 4)})
    return pd.DataFrame(pts)

if RUN_DISCOVERY:
    SCAN = build_scan_points()
    print(f"Scanning {len(SCAN)} pre-registered grid points across {len(SCAN_REGIONS)} regions…")
    rows = []
    for _, sp in SCAN.iterrows():
        try: f = fetch_power_features(sp.Scan_ID, sp.Lat, sp.Lon)
        except Exception as e: f = {"POWER_Status": f"Failed: {e}"}
        f["Scan_ID"] = sp.Scan_ID; rows.append(f)
        if f.get("POWER_Source") != "cache": time.sleep(0.35)
    SCAN_POWER = pd.DataFrame(rows).set_index("Scan_ID")

    plan = [(sp.Scan_ID, pt) for _, sp in SCAN.iterrows()
            for pt in grid_nodes(sp.Lat, sp.Lon, 3, DEM["primary"]["step_deg"], DEM["primary"]["step_deg"])]
    elev = []
    for i in range(0, len(plan), 100):
        chunk = plan[i:i + 100]
        data, _ = get_json(DEM_URL, {"latitude": ",".join(str(p[1][0]) for p in chunk),
                                     "longitude": ",".join(str(p[1][1]) for p in chunk)}, f"dem_scan_{i // 100:02d}")
        elev += [np.nan if z is None else float(z) for z in data.get("elevation", [np.nan] * len(chunk))]
    g = {}
    for (sid, _), z in zip(plan, elev): g.setdefault(sid, []).append(z)
    dem_rows = []
    for _, sp in SCAN.iterrows():
        Z = np.array(g.get(sp.Scan_ID, [np.nan] * 9)).reshape(3, 3)
        dem_rows.append({"Scan_ID": sp.Scan_ID, **(morph3(Z, sp.Lat) if np.isfinite(Z).all() else {})})
    SCAN_DEM = pd.DataFrame(dem_rows).set_index("Scan_ID")

    SF = SCAN.set_index("Scan_ID").join(SCAN_POWER).join(SCAN_DEM)
    ok = SF[SCORED_FEATURES].notna().all(axis=1)
    print(f"Complete scan points: {int(ok.sum())}/{len(SF)} (incomplete ones are excluded, never imputed)")
    X_SCAN = SF.loc[ok, SCORED_FEATURES].astype(float)

    SCAN_MOON, _  = score_against_reference(X_SCAN, LUNAR_REF)   # v8.1: same z-scale as the primary ranking
    SCAN_MARS, _  = score_against_reference(X_SCAN, MARS_REF)
    SCAN_POLAR, _ = score_against_reference(X_SCAN, POLAR_REF)
    SCAN_SCORED = SF.loc[ok, ["Region", "Region_Kind", "Lat", "Lon",
                              "Bare_Earth_Albedo", "Local_Relief_m", "Precipitation_mm_day"]].copy()
    SCAN_SCORED["Moon_Composite_%"] = SCAN_MOON["Composite_Score_%"]
    SCAN_SCORED["Moon_Geology_%"]   = SCAN_MOON["Geology_Index_%"]
    SCAN_SCORED["Mars_Composite_%"] = SCAN_MARS["Composite_Score_%"]
    SCAN_SCORED["Polar_Composite_%"] = SCAN_POLAR["Composite_Score_%"]
    SCAN_SCORED["Nearest_Known_Site_km"] = [
        min(haversine_km(r.Lat, r.Lon, k.Lat, k.Lon) for _, k in BLIND_REGISTRY.iterrows())
        for _, r in SCAN_SCORED.iterrows()]

    GEO_THRESHOLD = SCORES.loc[[x for x in PRIMARY if cat_p.get(x) == "Volcanic_Candidate"], "Geology_Index_%"].median()
    # (v8.1) threshold and scan scores are now on the SAME z-scale, so this comparison is meaningful
    SCAN_SCORED["Novel_Candidate"] = (SCAN_SCORED["Nearest_Known_Site_km"] >= NOVELTY_RADIUS_KM) &                                      (SCAN_SCORED["Moon_Geology_%"] >= GEO_THRESHOLD)
    SCAN_SCORED = SCAN_SCORED.sort_values("Moon_Composite_%", ascending=False)
    NOVEL = SCAN_SCORED[SCAN_SCORED["Novel_Candidate"]]
    print(f"\nPre-registered rule: {NOVELTY_RULE}")
    print(f"Geology threshold (median of primary volcanic candidates) = {GEO_THRESHOLD:.1f}%")
    print(f"→ {len(NOVEL)} novel candidate point(s) flagged for Phase-2 verification")
    sandk = SCAN_SCORED[SCAN_SCORED.Region_Kind == "sand_control"]
    volck = SCAN_SCORED[SCAN_SCORED.Region_Kind == "volcanic"]
    if len(sandk) and len(volck):
        print(f"Scan-level sanity: volcanic-region median Moon-Geology {volck['Moon_Geology_%'].median():.1f}% "
              f"vs sand-region {sandk['Moon_Geology_%'].median():.1f}%")
    display(SCAN_SCORED.head(12).round(2))
else:
    SCAN_SCORED = None; NOVEL = pd.DataFrame(); print("Discovery scan skipped (RUN_DISCOVERY = False).")
```

## 6g · Official challenge resources: live, keyless inventory

These cells query the resources on the challenge's Resources tab that need no account, and record **what each source actually has** over our Earth sites and over the two planetary targets. Every query is best-effort: a failed call is logged with its reason and never stops the notebook.

**Why SAR?** The lunar conduit under the Mare Tranquillitatis Pit was found with LRO's **Mini-RF radar**. RADARSAT-1 (CSA) and Sentinel-1 (ESA) are the Earth-side radars, so counting their coverage over each lava field is the first step toward comparing like with like: radar on both worlds.

```python
# ============================================================
# Cell 13g — Official resources inventory (Earthdata/CMR, Worldview, PDS ODE, ASF RADARSAT/S1, Copernicus)
# ============================================================
RESOURCE_LOG = []
def log_res(resource, agency, target, status, detail="", url=""):
    RESOURCE_LOG.append({"Resource": resource, "Agency": agency, "Target": target, "Status": status, "Detail": detail, "URL": url})

def bbox(lat, lon, d=0.02):
    return (round(lon - d, 4), round(lat - d, 4), round(lon + d, 4), round(lat + d, 4))   # W,S,E,N

PRIMARY_TABLE = BLIND_REGISTRY.set_index("Site_ID").loc[PRIMARY]
RESOURCE_LINKS = {}
for sid, r in PRIMARY_TABLE.iterrows():
    W, S, E, N = bbox(r.Lat, r.Lon, 0.25)
    RESOURCE_LINKS[sid] = {
        "earthdata_search": f"https://search.earthdata.nasa.gov/search?q=SRTMGL1&sb={W}%2C{S}%2C{E}%2C{N}",
        "worldview": f"https://worldview.earthdata.nasa.gov/?v={W},{S},{E},{N}&l=MODIS_Terra_CorrectedReflectance_TrueColor",
    }

if RUN_RESOURCE_INVENTORY:
    # ---- 1. NASA Earthdata Search back-end (CMR): granule counts per site ----
    CMR_URL = "https://cmr.earthdata.nasa.gov/search/granules.json"
    CMR_COLLECTIONS = {"SRTMGL1": "SRTM 30 m DEM", "HLSL30": "HLS Landsat 30 m reflectance",
                       "EMITL2BMIN": "EMIT L2B minerals", "ECO_L2T_LSTE": "ECOSTRESS 70 m land-surface temperature",
                       "OPERA_L2_RTC-S1_V1": "OPERA Sentinel-1 SAR backscatter"}
    cmr_rows = []
    for sid, r in PRIMARY_TABLE.iterrows():
        row = {"Site_ID": sid}
        for short, label in CMR_COLLECTIONS.items():
            try:
                resp = SESSION.get(CMR_URL, params={"short_name": short, "bounding_box": ",".join(map(str, bbox(r.Lat, r.Lon))),
                                                    "page_size": 0}, timeout=40)
                hits = resp.headers.get("CMR-Hits") if hasattr(resp, "headers") else None
                row[short] = int(hits) if hits is not None else np.nan
            except Exception as e:
                row[short] = np.nan
        cmr_rows.append(row); time.sleep(0.2)
    CMR_INVENTORY = pd.DataFrame(cmr_rows).set_index("Site_ID")
    ok_cmr = CMR_INVENTORY.notna().any(axis=1).sum()
    log_res("NASA Earthdata Search (CMR)", "NASA", "8 Earth sites", "live" if ok_cmr else "unavailable here",
            f"granule counts for {', '.join(CMR_COLLECTIONS)} at {ok_cmr}/8 sites", "https://search.earthdata.nasa.gov/")

    # ---- 2. NASA Worldview / GIBS: tile reachability ----
    GIBS_TILE = ("https://gibs.earthdata.nasa.gov/wmts/epsg3857/best/MODIS_Terra_CorrectedReflectance_TrueColor/default/"
                 "2024-07-01/GoogleMapsCompatible_Level9/{z}/{y}/{x}.jpg")
    try:
        t = SESSION.get(GIBS_TILE.format(z=0, y=0, x=0), timeout=30)
        log_res("NASA Worldview (GIBS)", "NASA", "map + 8 site links", "live" if getattr(t, "status_code", 0) == 200 else "unavailable here",
                "MODIS Terra true-colour layer added to the map; Worldview link per site", "https://worldview.earthdata.nasa.gov/")
    except Exception as e:
        log_res("NASA Worldview (GIBS)", "NASA", "map + 8 site links", "unavailable here", type(e).__name__, "https://worldview.earthdata.nasa.gov/")

    # ---- 3. NASA PDS (ODE REST): LRO products over the lunar pit, MRO products over Arsia ----
    ODE = "https://oderest.rsl.wustl.edu/live2/"
    def ode_count(target, ihid, iid, lat, lon, d=0.05):
        lon360 = lon % 360
        q = {"query": "product", "results": "c", "output": "JSON", "target": target, "ihid": ihid, "iid": iid,
             "minlat": lat - d, "maxlat": lat + d, "westernlon": lon360 - d, "easternlon": lon360 + d}
        resp = SESSION.get(ODE, params=q, timeout=60)
        js = resp.json()
        return int(js.get("ODEResults", {}).get("Count", js.get("Count")))
    PDS_INV = []
    for tgt, ihid, iid, lat, lon, label in [
            ("moon", "LRO", "LROC", LUNAR_SITE["lat"], LUNAR_SITE["lon"], "LROC images over the Mare Tranquillitatis Pit"),
            ("moon", "LRO", "DIVINER", LUNAR_SITE["lat"], LUNAR_SITE["lon"], "Diviner products over the pit"),
            ("moon", "LRO", "MINI-RF", LUNAR_SITE["lat"], LUNAR_SITE["lon"], "Mini-RF radar products over the pit"),
            ("mars", "MRO", "HIRISE", MARS_SITE["lat"], MARS_SITE["lon"], "HiRISE images over the Arsia skylight region"),
            ("mars", "MRO", "CTX", MARS_SITE["lat"], MARS_SITE["lon"], "CTX images over the Arsia skylight region")]:
        try:
            n = ode_count(tgt, ihid, iid, lat, lon); PDS_INV.append({"Target": tgt, "Instrument": f"{ihid}/{iid}", "Products": n, "What": label})
        except Exception as e:
            PDS_INV.append({"Target": tgt, "Instrument": f"{ihid}/{iid}", "Products": np.nan, "What": label + f" (query failed: {type(e).__name__})"})
        time.sleep(0.3)
    PDS_INV = pd.DataFrame(PDS_INV)
    log_res("NASA PDS (ODE) + LRO/MRO", "NASA", "Moon pit + Arsia skylights",
            "live" if PDS_INV.Products.notna().any() else "unavailable here",
            f"{int(PDS_INV.Products.notna().sum())}/{len(PDS_INV)} instrument queries answered", "https://ode.rsl.wustl.edu/")

    # ---- 4. CSA RADARSAT-1 + ESA Sentinel-1 SAR archive via NASA ASF (radar ↔ LRO Mini-RF) ----
    ASF = "https://api.daac.asf.alaska.edu/services/search/param"
    sar_rows = []
    for sid, r in PRIMARY_TABLE.iterrows():
        row = {"Site_ID": sid}
        for plat in ["RADARSAT-1", "SENTINEL-1"]:
            try:
                resp = SESSION.get(ASF, params={"platform": plat, "intersectsWith": f"POINT({r.Lon} {r.Lat})", "output": "count"}, timeout=60)
                row[plat] = int(str(resp.text).strip())
            except Exception:
                row[plat] = np.nan
        sar_rows.append(row); time.sleep(0.2)
    SAR_INVENTORY = pd.DataFrame(sar_rows).set_index("Site_ID")
    ok_sar = SAR_INVENTORY.notna().any(axis=1).sum()
    log_res("CSA RADARSAT-1 (+ ESA Sentinel-1) via NASA ASF", "CSA / ESA", "8 Earth sites",
            "live" if ok_sar else "unavailable here", f"SAR scene counts at {ok_sar}/8 sites — Earth counterpart of LRO Mini-RF",
            "https://search.asf.alaska.edu/")

    # ---- 5. ESA Copernicus Data Space: Sentinel-2 coverage ----
    CDSE = "https://catalogue.dataspace.copernicus.eu/odata/v1/Products"
    cdse_ok = 0; CDSE_COUNTS = {}
    for sid, r in PRIMARY_TABLE.iterrows():
        try:
            flt = (f"Collection/Name eq 'SENTINEL-2' and OData.CSC.Intersects(area=geography'SRID=4326;POINT({r.Lon} {r.Lat})')")
            resp = SESSION.get(CDSE, params={"$filter": flt, "$count": "True", "$top": 1}, timeout=60)
            CDSE_COUNTS[sid] = int(resp.json().get("@odata.count")); cdse_ok += 1
        except Exception:
            CDSE_COUNTS[sid] = np.nan
        time.sleep(0.2)
    log_res("ESA Copernicus Data Space Ecosystem", "ESA", "8 Earth sites", "live" if cdse_ok else "unavailable here",
            f"Sentinel-2 product counts at {cdse_ok}/8 sites; Copernicus DEM is the core elevation source", "https://dataspace.copernicus.eu/")
else:
    CMR_INVENTORY = SAR_INVENTORY = pd.DataFrame(); PDS_INV = pd.DataFrame(); CDSE_COUNTS = {}

# ---- 6. Account-only resources: documented, not silently dropped ----
log_res("CSA RADARSAT Constellation Mission (RCM)", "CSA", "Phase 2", "link only", "EODMS account required", "https://www.eodms-sgdot.nrcan-rncan.gc.ca/")
log_res("CSA SAR Python tutorial", "CSA", "method", "reference", "basis for §9c backscatter cell", "https://www.asc-csa.gc.ca/")
log_res("ESA Open Science Catalog", "ESA", "Phase 2", "link only", "browse catalog for geoscience layers", "https://opensciencedata.esa.int/")
log_res("JAXA G-Portal", "JAXA", "Phase 2", "link only", "account required; GSMaP precipitation / GCOM-C albedo cross-check", "https://gportal.jaxa.jp/")
log_res("NASA LRO Diviner Lunar Data", "NASA", "Moon pit + lunar south pole", "used",
        "pit/surface temperatures in §6c; polar temperatures define the v8.1 south-pole target and cold-context rule (§6e2)", "https://www.diviner.ucla.edu/")

RESOURCE_REPORT = pd.DataFrame(RESOURCE_LOG)
print("OFFICIAL RESOURCES — what this run actually reached")
display(RESOURCE_REPORT[["Resource", "Agency", "Target", "Status", "Detail"]])
if RUN_RESOURCE_INVENTORY:
    if len(CMR_INVENTORY): print("\nEarthdata (CMR) granules per site:"); display(CMR_INVENTORY)
    if len(SAR_INVENTORY): print("\nSAR scenes per site (RADARSAT-1 = CSA, Sentinel-1 = ESA; via NASA ASF):"); display(SAR_INVENTORY)
    if len(PDS_INV): print("\nPDS products over the planetary targets:"); display(PDS_INV)
```

## 7b · Multi-criteria suitability matrix and heat-maps (v8.1)

A heat-map is only honest if every cell is on the same scale. Here each cell is the per-feature similarity `100·exp(−k·|Δz|)` under the **frozen reference scaling**, for every site (primary + benchmark) × every feature × every target. The right-hand column of each panel is the composite. Dark = far from the target. Read across a row to see *why* a site scores what it scores; read down a column to see which feature separates the sites.

Two things the heat-map cannot say, printed on the figure: the albedo cell is a 0.5–1° POWER pixel (so El Pinacate's cell is dune-contaminated), and no cell measures a cave.

```python
# ============================================================
# Cell 14b — Multi-criteria suitability matrix + heat-maps (v8.1)
# ============================================================
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
MC_IDS = [s for s in RANKED.index] + [s for s in BENCH]
X_MC = FT.loc[MC_IDS, SCORED_FEATURES].astype(float)
FEATS = list(NICE)                                           # 7 scored features, display order from §7
mc_rows = []
for t, (ref, _site) in TARGETS.items():
    sc_t, dz_t = score_against_reference(X_MC, ref)
    for sid in MC_IDS:
        row = {f: 100 * math.exp(-DECAY[PILLAR_OF[f]] * abs(dz_t.at[sid, f])) for f in FEATS}
        row["Composite_%"] = sc_t.at[sid, "Composite_Score_%"]; row["Geology_%"] = sc_t.at[sid, "Geology_Index_%"]
        mc_rows.append({"Site_ID": sid, "Target": t, **row})
MULTI_CRITERIA = pd.DataFrame(mc_rows).set_index(["Site_ID", "Target"])

CMAP = LinearSegmentedColormap.from_list("cs", ["#111827", "#7c2d12", "#f59e0b", "#fef3c7"])
labels_y = [f"{FT.at[s, 'Site_Name'].split(' (')[0]}{' ◆' if s in BENCH else ''}" for s in MC_IDS]
labels_x = [NICE[f][0] for f in FEATS] + ["COMPOSITE"]
fig, axes = plt.subplots(1, 3, figsize=(21, 0.52 * len(MC_IDS) + 2.6), sharey=True)
for ax, t in zip(axes, TARGETS):
    M = MULTI_CRITERIA.xs(t, level="Target").loc[MC_IDS, FEATS + ["Composite_%"]].values
    im = ax.imshow(M, cmap=CMAP, vmin=0, vmax=100, aspect="auto")
    ax.set_xticks(range(len(labels_x)), labels_x, rotation=40, ha="right", fontsize=8)
    ax.set_yticks(range(len(MC_IDS)), labels_y, fontsize=8)
    for i in range(M.shape[0]):
        for j in range(M.shape[1]):
            ax.text(j, i, f"{M[i, j]:.0f}", ha="center", va="center", fontsize=7,
                    color="#111827" if M[i, j] > 55 else "#f9fafb", fontweight="bold" if j == M.shape[1] - 1 else "normal")
    ax.axvline(len(FEATS) - 0.5, color="white", lw=2)
    ax.set_title(f"Target: {TARGETS[t][1]['name']}", fontsize=10)
fig.colorbar(im, ax=axes, fraction=0.012, pad=0.01, label="% similarity to target (frozen reference z-scale)")
fig.suptitle("CAVE-SIGHT multi-criteria suitability matrix — site × feature × target   (◆ = benchmark analog, scored not tested)", fontsize=12, y=1.01)
fig.text(0.01, -0.02, "Caveats: albedo/climate cells come from 0.5°–1° NASA POWER pixels (dune mixing at El Pinacate); relief/slope from a 300 m GLO-90 window; "
         "no cell measures a cave. Polar 'cold' is reported separately (§6e2), not inside the score.", fontsize=8, color="#6b7280")
fig.savefig(OUT_DIR / "cavesight_suitability_heatmap.png", dpi=150, bbox_inches="tight"); plt.show()

# --- compact site × target composite heat-map (the "which world does this place resemble" view) ---
ST = MULTI_CRITERIA["Composite_%"].unstack("Target").loc[MC_IDS, list(TARGETS)]
fig, ax = plt.subplots(figsize=(6.2, 0.42 * len(MC_IDS) + 1.6))
im = ax.imshow(ST.values, cmap=CMAP, vmin=0, vmax=100, aspect="auto")
ax.set_xticks(range(3), ["Moon pit", "Mars skylights", "Lunar S. pole"]); ax.set_yticks(range(len(MC_IDS)), labels_y, fontsize=8)
for i in range(ST.shape[0]):
    for j in range(ST.shape[1]):
        ax.text(j, i, f"{ST.values[i, j]:.0f}", ha="center", va="center", fontsize=8, color="#111827" if ST.values[i, j] > 55 else "#f9fafb")
ax.set_title("Composite similarity by target (%)", fontsize=10); fig.colorbar(im, ax=ax, fraction=0.05)
fig.savefig(OUT_DIR / "cavesight_site_by_target_heatmap.png", dpi=150, bbox_inches="tight"); plt.show()
print("Saved: cavesight_suitability_heatmap.png · cavesight_site_by_target_heatmap.png")
```

```python
# ============================================================
# Cell 18b — 9c · OPERA Sentinel-1 RTC radar backscatter per site (optional, Earthdata login)
#   Earth-side radar measurement to pair with LRO Mini-RF. Method follows the CSA SAR Python tutorial
#   (open a calibrated SAR raster, convert to dB, sample a window). Not executed in the build sandbox.
# ============================================================
if RUN_SAR_BACKSCATTER:
    !pip -q install earthaccess rasterio pyproj
    import earthaccess, rasterio
    from pyproj import Transformer
    earthaccess.login(strategy="interactive", persist=True)
    sar = []
    for sid, r in BLIND_REGISTRY.set_index("Site_ID").loc[PRIMARY].iterrows():
        res = earthaccess.search_data(short_name="OPERA_L2_RTC-S1_V1", bounding_box=(r.Lon - .01, r.Lat - .01, r.Lon + .01, r.Lat + .01), count=1)
        if not res: sar.append({"Site_ID": sid, "SAR": "no OPERA RTC granule"}); continue
        links = [l for l in res[0].data_links() if l.endswith("_VV.tif")]
        if not links: sar.append({"Site_ID": sid, "SAR": "no VV band"}); continue
        f = earthaccess.open([links[0]])[0]
        with rasterio.open(f) as ds:
            x, y = Transformer.from_crs("EPSG:4326", ds.crs, always_xy=True).transform(r.Lon, r.Lat)
            row, col = ds.index(x, y)
            win = ds.read(1, window=((row - 2, row + 3), (col - 2, col + 3))).astype(float)
        win[win <= 0] = np.nan
        sar.append({"Site_ID": sid, "SAR": "ok", "VV_backscatter_dB": float(10 * np.log10(np.nanmean(win))),
                    "Granule": res[0]["umm"]["GranuleUR"]})
    SAR_DB = pd.DataFrame(sar).set_index("Site_ID").join(LABELS["PostHoc_Category"])
    SAR_DB.to_csv(OUT_DIR / "cavesight_sar_backscatter.csv"); display(SAR_DB.round(2))
    print("Read: rough basalt usually backscatters more strongly than smooth sand. This is a descriptive layer, not a tube detector.")
else:
    print("Skipped (RUN_SAR_BACKSCATTER = False).")
```

## 10 · How to report the results honestly

**What the pipeline can support**
- How close each site's *measurable surface proxies* come to the lunar reference, with uncertainty (Monte Carlo) and robustness (specification curve).
- Which factor drives each site's score: geology or climate.

**What it cannot support**
- That a site contains a lava tube, or that its conduit is any particular width. A DEM sees only the surface. The 45 m match is a literature finding.
- That Umm Jirsan's #1 rank *validates* the method. The climate targets describe a hot, dark, hyper-arid desert, and Umm Jirsan is the only site that is both dark and hyper-arid.
- A strong statistical claim from 5 vs 2 sites. Even perfect separation gives only p ≈ 0.048.

**Known limitations → Phase 2 fixes**
- NASA POWER pixels are 55–110 km and mix basalt with nearby sand (El Pinacate). Fix: 30 m HLS albedo and 60 m EMIT L2B mineralogy (§9b).
- The climate targets are terrestrial proxies. v8.1 adds the lunar **south-pole target and four benchmark analogs (two of them polar)** at proxy level (§6e2); the cold itself is a reported descriptor, not a scored feature. Phase 2 replaces the proxies with Diviner-measured polar temperatures, LEND hydrogen, and LOLA illumination maps for the Shackleton rim.
- The benchmark analogs are **scored, not tested**: with one site per environment there is no control group, so their numbers describe, they do not prove.
- More sites with **rule-based** selection would add statistical power. Example: every lava-tube field in a published catalogue, plus controls sampled at random from sand-sea polygons.
- **Mars:** Mars Trek skylights at Arsia Mons and Olympus Mons (g = 3.71 m s⁻²). **Roof stability:** GSI 50–70 beam/arch models across 9.81 / 3.71 / 1.62 m s⁻².
